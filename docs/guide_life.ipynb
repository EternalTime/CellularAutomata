## B/S notation

Life is one member of the outer-totalistic family: rules that update a cell from its own value and the *sum* of its eight Moore neighbors, nothing more.
The family is named in B/S notation - B lists the neighbor counts at which a dead cell is born, S the counts at which a live cell survives - making Life B3/S23.
The [`pyCA.ca2d.CA2D`](#pyCA.ca2d.CA2D) class takes the two lists directly, and Life gets a named constructor:

````python
from pyCA import CA2D

life = CA2D.life((128, 128))      # random soup at density 0.5
life.run(200)
print(life.population)
````

````python
life.play()                       # watch it live; close to stop
````

We keep the whole history of one soup and look at a few snapshots of it.
Within ten steps the random start has thinned to small clusters, and by step 1000 only a sparse scatter of still lifes and blinkers remains.

````python
import numpy as np
import matplotlib.pyplot as plt
from pyCA import CA2D

life = CA2D.life((64, 64), memory=1001, rng=np.random.default_rng(2))
life.run(1000)
history = life.history()          # every state from t = 0 to t = 1000

times = [0, 10, 100, 1000]
fig, axes = plt.subplots(1, len(times), figsize=(6.4, 1.8), constrained_layout=True)
for t, ax in zip(times, axes):
    ax.imshow(history[t], cmap='binary', interpolation='nearest')
    ax.set_title('$t = %d$' % t)
    ax.set_xticks([])
    ax.set_yticks([])
plt.show()
````

Summing each snapshot gives the number of live cells, and we plot that number against time.
The population falls by more than half in the first ten steps, declines slowly as the surviving clusters collide, and levels off after about 600 steps, once only the still lifes and blinkers are left.

````python
plt.plot(history.sum(axis=(1, 2)))
plt.xlabel('time step')
plt.ylabel('live cells')
plt.show()
````

The lattice is periodic in both directions, so gliders that leave one edge arrive at the other.

## Building from a known pattern

Pass an explicit state to study a specific object.
The glider, the smallest spaceship, translates one cell diagonally every four steps - the repository's test suite holds the implementation to exactly that:

````python
import numpy as np
from pyCA import CA2D

state = np.zeros((40, 40), dtype=int)
for r, c in [(0, 1), (1, 2), (2, 0), (2, 1), (2, 2)]:
    state[r + 18, c + 18] = 1

life = CA2D.life(state=state)
life.run(4)        # the glider has moved by (1, 1)
````

## Beyond Life

Change the two lists and the physics changes with them.
HighLife (B36/S23) adds a single birth condition to Life and gains a replicator.
Seeds (B2/S) is pure birth - nothing survives, yet the explosions are intricate.
Day & Night (B3678/S34678) is symmetric under exchanging live and dead cells, so every pattern has a photographic negative with identical dynamics.

````python
highlife = CA2D([3, 6], [2, 3], shape=(128, 128))
seeds    = CA2D([2], [], shape=(128, 128), fill=0.05)
daynight = CA2D([3, 6, 7, 8], [3, 4, 6, 7, 8], shape=(128, 128))
````

Start HighLife and Seeds from the same random soup and run each for 100 steps.
HighLife thins to scattered clusters much as Life does, while Seeds keeps the whole lattice churning: every live cell dies on the next step, and births replace them.

````python
highlife = CA2D([3, 6], [2, 3], shape=(128, 128), rng=np.random.default_rng(3))
seeds    = CA2D([2], [], shape=(128, 128), rng=np.random.default_rng(3))

fig, axes = plt.subplots(1, 2, figsize=(6.4, 3.4), constrained_layout=True)
for name, ca, ax in [('HighLife', highlife, axes[0]), ('Seeds', seeds, axes[1])]:
    ca.run(100)
    ax.imshow(ca.state, cmap='binary', interpolation='nearest')
    ax.set_title(name)
    ax.set_xticks([])
    ax.set_yticks([])
plt.show()
````

Run a random soup under each and compare the late-time populations.
Which rules die, which saturate, and which hover?
